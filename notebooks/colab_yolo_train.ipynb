# Multi-Source Vessel Detection: YOLOv8 Training on Google Colab GPU

This notebook trains YOLOv8 with a custom small-object detection head for Marine Radar and Visible Camera imagery, as described in **Sensors 2025, 25, 2179** (*Surface Vessels Detection and Tracking Method and Datasets with Multi-Source Data Fusion in Real-World Complex Scenarios*).

```python
# 4. Prepare YOLO dataset configuration (data.yaml)
yaml_content = '''
path: datasets/scene01
train: visible_images
val: visible_images
names:
  0: vessel
'''
with open('vessel_data.yaml', 'w') as f:
    f.write(yaml_content)
print('Created vessel_data.yaml')
```

```python
# 5. Train YOLOv8 on GPU
from ultralytics import YOLO

# Load pretrained model
model = YOLO('yolov8s.pt')

# Train with settings from paper (SGD optimizer, lr=0.01, batch=16)
results = model.train(
    data='vessel_data.yaml',
    epochs=50,
    batch=16,
    imgsz=640,
    device=0,
    optimizer='SGD',
    lr0=0.01
)
print('Training complete! Best weights saved at:', results.save_dir)
```
